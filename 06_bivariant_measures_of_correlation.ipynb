## Task 1

```python
from ucimlrepo import fetch_ucirepo 

# fetch dataset 
cars = fetch_ucirepo(id=9) 
# https://archive.ics.uci.edu/dataset/9/auto+mpg

# data (as pandas dataframes) 
data = cars.data.features

# Show the first 5 rows
data.head()
```

1. Analyze the correlation between the variables horsepower and cylinders. Therefore calculate the covariance, pearson correlation coefficient and spearman correlation coefficient. Interpret the results.

```python
import pandas as pd

# covariance
cov = data['horsepower'].cov(data['cylinders'], ddof=0)
print(f"Covariance: {cov}")

# pearson correlation coefficient
pearson = data['horsepower'].corr(data['cylinders'],method='pearson')
print(f"Pearson Correlation Coefficient: {pearson}")

# spearman correlation coefficient
spearman = data['horsepower'].corr(data['cylinders'],method='spearman')
print(f"Spearman Correlation Coefficient: {spearman}")
```

#### Interpretation:
- Kovarianz: 55,2 > 0: Autos mit mehr Zylindern haben mehr PS
- Pearson r = 0,84: starken positiven linearen Zusammenhang weil r nahe bei 1 ist
- Spearman r = 0,82: starker positiver Zusammenhang


2. Generate a scatter plot for the variabels horsepower and cylinders. Compare the before result with the calculated measures.

```python
import pandas as pd
import plotly.express as px

# Create a scatter plot
fig = px.scatter(data, x="horsepower", y="cylinders")

# Adjust the plot
fig.update_layout(
    xaxis_title_text='Horsepower',
    yaxis_title_text='Cylinders',
    title=dict(
            text='Scatter Plot',
        ),
)

# Show the plot
fig.show()
```

## Task 2

```python
import pandas as pd

income = [150, 250, 175, 165]
expenditure = [135, 150, 140, 150]

# convert the lists to pandaframe
df = pd.DataFrame({'income': income, 'expenditure': expenditure})
df.head()
```

1. Calculate the covariance between income and expenditure. Interpret the result.

```python
import pandas as pd

# Calculate covariance
cov_euro = df['income'].cov(df['expenditure'], ddof=0) # ddof=0 --> Formlar for sample | ddof=1 --> Formlar for population (default)
print(f"Covariance: {cov_euro}")

#interpretieren!!!
```

2. Calculate the covariance when income is measured in Euro cents. How does this affect the interpretation?

```python
# Calculate covariance
cov_cents = (df['income']*100).cov(df['expenditure'], ddof=0) # ddof=0 --> Formlar for sample | ddof=1 --> Formlar for population (default)
print(f"Covariance: {cov_cents}")

#interpretieren!!!
```

3. Switch back to income in Euro. Calculate the Pearson correlation coefficient. Interpret the results.

```python
# Pearson correlation coefficient
pearson_euro = df['income'].corr(df['expenditure'],method='pearson')
print(f"Pearson Correlation Coefficient: {pearson_euro}")

#interpretieren!!!
```
