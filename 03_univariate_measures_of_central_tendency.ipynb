# Measures of Central Tendency

```python
from ucimlrepo import fetch_ucirepo 

# fetch dataset 
cars = fetch_ucirepo(id=9) 
# https://archive.ics.uci.edu/dataset/9/auto+mpg

# data (as pandas dataframes) 
data = cars.data.features
data = data.join(cars.data.ids)

# Show the first 5 rows
data.head()
```

#### Calculate all useful and meaningful measure of central tendency (mean, median, mode) for the following attributes (think about attribute types)
- car_name
- origin
- displacement

```python
import statistics


print('Mode car_name: ' + str(statistics.mode(data['car_name']))) 
print('Mode origin: ' + str(statistics.mode(data['origin']))) 
print('Mode displacement: ' + str(statistics.mode(data['displacement']))) 

print('---')

print('Mean displacement: ' + str(statistics.mean(data['displacement'])))
print('Median displacement: ' + str(statistics.median(data['displacement'])))

```

#### Calculate the quartiles and plot a Boxplot fo the attribute acceleration

```python
import numpy as np
from scipy import stats

# Quantiles:
print(stats.mstats.mquantiles(data['acceleration'], prob=[0.25, 0.5, 0.75]))

# Boxplot:
import plotly.express as px
import pandas as pd

fig = px.box(pd.DataFrame(data['acceleration']))

fig.update_layout(
    xaxis_title_text=' ',
    yaxis_title_text='Acceleration',
)

fig.show()

```
