## Task 1

```python
from ucimlrepo import fetch_ucirepo 

# fetch dataset 
cars = fetch_ucirepo(id=9) 
# https://archive.ics.uci.edu/dataset/9/auto+mpg

# data (as pandas dataframes) 
data = cars.data.features
data = data.join(cars.data.ids)

# Show the first 5 rows
data.head()
```

#### For the attribute acceleration calculate the following measures (use the sample formula - not population):
- Range
- IQR --> compare to the boxplot from the section Measures of Central Tendency
- Variance
- Standard Deviation
- CV

```python
import numpy as np
from scipy import stats
import statistics 

# Range
print('Range:', np.ptp(data['acceleration']))

# IQR
print('IQR: ', stats.iqr(data['acceleration'], interpolation = 'nearest'))

# Variance
print('Variance: ', statistics.variance(data['acceleration']))

# Standard Deviation
print('Standard Deviation: ', statistics.stdev(data['acceleration']))

# Coefficient of Variation
print('CV: ' + str(stats.variation(data['acceleration'])))
```

## Task 2

```python
# Website: https://jse.amstat.org/v14n2/datasets.aerts.html
# Dataset: https://jse.amstat.org/datasets/euroweight.dat.txt
# Description: https://jse.amstat.org/datasets/euroweight.txt

import pandas as pd
import numpy as np

# Load the dataset
data = pd.read_csv('Daten/euroweight.dat.txt', sep='\t', header=None, index_col=0, names=['Weight', 'Batch'])

# Display the first few rows
data.head()
```

As the Head of Quality Control at the European Central Bank (ECB), you are responsible, among other duties, for the quality management of 1-Euro coins. Consequently, you have tasked an employee with selecting a random sample of 2,000 coins. (Dataset: 'euro.csv', Unit: grams)

- Calculate the average weight of the coins.
- Determine the corresponding standard deviation and interpret its significance.
- Create a histogram. Ensure that all axes are labeled and the chart is properly titled.

```python
import statistics

# average weight of the coins
print('Average weight of the coins: ' + str(statistics.mean(data['Weight'])))

# standard deviation
print('Standard Deviation: ', statistics.stdev(data['Weight']))

# histogram
import plotly.express as px

# Generate Histogram
fig = px.histogram(
    data, 
    x="Weight",
    nbins=50,
)
# Adjust the plot
fig.update_layout(
    title=dict(
            text='Histogram of 1-Euro-Coins',
        ),
    xaxis_title_text='Weight',
    yaxis_title_text='Amount of coins',
)
# Show the plot
fig.show()

```
