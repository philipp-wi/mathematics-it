# Frequency Distribution

```python
from ucimlrepo import fetch_ucirepo 

# fetch dataset 
cars = fetch_ucirepo(id=9) 
# https://archive.ics.uci.edu/dataset/9/auto+mpg

# data (as pandas dataframes) 
data = cars.data.features
data = data.join(cars.data.ids)

# Show the first 5 rows
data.head()
```

#### Histogram | Absolute Frequency | Variable: origin

```python
import plotly.express as px

fig = px.histogram(
    data, 
    x="origin",
)

fig.update_layout(
    xaxis_title_text='Origin',
    yaxis_title_text='Absolute Frequency',
    title="Absolute Frequency of car origin",
)

# Show the plot
fig.show()
```

#### Bar Chart | Absoulte Frequency | no binning | Variable: weight

#### Histogram | Absoulte Frequency | automatic binning | Variable: weight

```python
import plotly.express as px

fig = px.histogram(
    data,
    x="weight",
)

fig.update_layout(
    xaxis_title_text='Weight',
    yaxis_title_text='Absolute Frequency',
    title=dict(
            text='Automatic Binning Histogram',
        ),
)

fig.show()
```

#### Histogram | Relative Frequency | cumulated | Variable: hoursepower

```python
import plotly.express as px
# Generate Histogram
fig = px.histogram(
    data, 
    x="horsepower",
    cumulative=True,
    histnorm="percent"
)

# Adjust the plot
fig.update_layout(
    xaxis_title_text='horsepower',
    yaxis_title_text='Relative Frequency [%]',
    title=dict(
            text='<b><span style="font-size: 10pt">Ordinal Variable: Cumulated</span> ',
        ),
)

# Show the plot
fig.show()
```

#### Pie Chart | Relative Frequency | Variable: cylinders

```python
import plotly.express as px

fig = px.pie(
    data,
    names="cylinders",
)

# Adjust the plot
fig.update_layout(
    title=dict(
            text='<b><span style="font-size: 10pt">Nominal Variable: Pie Chart</span>',
        ),
)
fig.update_traces(textposition='outside', textinfo='percent+label')

# Show the plot
fig.show()
```
