# Frequency Distribution

```python
from ucimlrepo import fetch_ucirepo 

# fetch dataset 
cars = fetch_ucirepo(id=9) 
# https://archive.ics.uci.edu/dataset/9/auto+mpg

# data (as pandas dataframes) 
data = cars.data.features

# Show the first 5 rows
data.head()
```

1. Generate a 2D Histogram for the variables origin and horsepower (think about attribute types, title, labeling of the axes). Interpret the results.

```python
import plotly.express as px

# Create 2D histogram
fig = px.density_heatmap(data, x="origin", y="horsepower")

# Adjust the plot
fig.update_layout(
    xaxis_title_text='Origin',
    yaxis_title_text='Horsepower',
    title=dict(
            text='2D Histogram of origin and horsepower',
        ),
)

# Show the plot
fig.show()
```

2. Calculate the crosstab for the absolute frequencies of the variables origin and cylinders

```python
import pandas as pd
import plotly.express as px

df = px.data.tips()
pd.crosstab( data['origin'], data['cylinders'], margins=True)
```
