## Task 2

Let's stick with the example from before and perform further experiments. We use a fair die and a biased die and perform the following task:

1. Roll each die
- 5
- 50
- 500 times.
2. Visualize the results in histogram (one for the fair die, one for the biased die).

```python
import random
import plotly.express as px

die_fair = [1, 2, 3, 4, 5, 6]
die_biased = [1, 2, 4, 5, 6, 6]

random.seed(23)

# Fair die
for n in [5, 50, 500]:
    rolls = random.choices(die_fair, k=n)

    fig = px.histogram(
        x=rolls,
        nbins=6,
        title=f"Fair Die - {n} rolls"
    )
    fig.show()

# Biased die
for n in [5, 50, 500]:
    rolls = random.choices(die_biased, k=n)

    fig = px.histogram(
        x=rolls,
        nbins=6,
        title=f"Biased Die - {n} rolls"
    )
    fig.show()
```

## Task 3

```python
import random
import pandas as pd
import plotly.express as px

die_fair = [1, 2, 3, 4, 5, 6]

random.seed(23)
rolls = random.choices(die_fair, k=500)

# Calculate probabilities
probability = pd.Series(die_fair).value_counts(normalize=True).sort_index()

# Create histogram
fig = px.histogram(x=rolls, nbins=6, histnorm='probability')

fig.update_layout(
    xaxis_title='Dice Number',
    yaxis_title='Probability/Proportion',
    title='Fair Die - 500 Rolls'
)

# Overlay theoretical probabilities
for i in range(1, 7):
    fig.add_shape(
        type='line',
        x0=i-0.46,
        x1=i+0.46,
        y0=probability[i],
        y1=probability[i],
        line=dict(color='#E87F2B', width=4)
    )

fig.show()

print(probability)
```

## Task 4

```python
# Fair die
p_fair = 1/6
odds_fair = p_fair / (1-p_fair)

# Biased die
p_biased = 2/6
odds_biased = p_biased / (1-p_biased)

print("Fair die odds:", odds_fair)
print("Biased die odds:", odds_biased)
```

## Task 5

Assume the heights of individuals in a certain population follow a normal distribution with a mean of 170 cm and a standard deviation of 10 cm (see examples above).

Answer the following questions based on this normal distribution:

1. What percentage of individuals are taller than 190 cm?
2. What percentage of individuals are between 170 cm and 180 cm tall?
3. Plot the Probability Density Function (PDF) for a normal distribution with a mean of 180 cm and a standard deviation of 5 cm.

For the first two questions use the Cumulative Distribution Function (CDF) to calculate the corresponding probabilities.

```python
from scipy.stats import norm

# 1.
print('Taller than 190cm:', (1 - norm.cdf(190, 170, 10))*100, '%')  
# print('Smaller than 150cm:', (norm.cdf(150, 170, 10))*100, '%')

# 2.
print('Between 170cm and 180cm:', (norm.cdf(180, 170, 10) - norm.cdf(170, 170, 10))*100, '%')

```

```python
# 3.
import numpy as np
import pandas as pd
import plotly.express as px
from scipy.stats import norm

# keine limits!

# Generate data for the normal distribution
x = np.arange(160, 200, 0.1)          
y = norm.pdf(x, 180, 5)    # (µ=180, std=5)
df = pd.DataFrame({'x': x, 'y': y})

# Create the plot
fig = px.line(df, x='x', y='y')

# Adjust the plot
fig.data[0].update(line=dict(color='#00416E', width=2))


fig.update_layout(
    xaxis_title_text='Height [cm]',  
    yaxis_title_text='Density',
    title=dict(
            text=f'<b><span style="font-size: 10pt">Probability Density Function </span><br> <span style="font-size:5">µ=180cm, std=5cm</span> </b>',  # ← geändert
        ),
    showlegend=False,
)

fig.show()
```
