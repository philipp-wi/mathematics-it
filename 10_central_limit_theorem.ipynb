# Central Limit Theorem

We use the biased die example from before:


die_biased = [1, 2, 4, 5, 6, 6]
Work on the following tasks:

1. Calculate the expected value of the biased die (mean).
2. Visualize the probability of each side in a histrogram.

3. Now we perform some experiments:
- Choose one random side (sample_size = 1) of the biased die. Repeat this 100.000 times and visualize the frequency in a histogram.
- Compare the result with the visualization of the probability.
- Now increase the sample_size and inspect the change in histogram.
- Does the calculated expected value match with the mean value of the normal distribution?

```python
import random
import numpy as np
import plotly.express as px

die_biased = [1, 2, 4, 5, 6, 6]

# 1.
expected_value = np.mean(die_biased)
print('Expected value of biased die:', expected_value)
```

```python
# 2.
import plotly.express as px

fig = px.histogram(x=die_biased, histnorm='probability')
fig.update_traces(xbins=dict(start=0.5, end=6.5, size=1),
                  marker=dict(color='rgba(0, 65, 110, 0.8)'))

fig.update_layout(
        barmode='overlay',
        yaxis_title_text='Probability',
        xaxis_title_text='Die Number',
        bargap=0.1,
        title=dict(
            text='<b><span style="font-size: 10pt">Probability</span><br> <span style="font-size:5">Data: Biased Die</span></b>',
        ),
    )

fig.show()
```

```python
#3. 
for sample_size in [1, 2, 5, 30]:

    sample_means = []
    for i in range(100000):
        sample_means.append(np.mean(random.choices(die_biased, k=sample_size)))

    fig = px.histogram(x=sample_means)
    fig.update_traces(marker=dict(color='rgba(0, 65, 110, 0.8)'))

    fig.add_vline(x=expected_value, line_dash="dash", annotation_text="Expected Value: "+str(expected_value), annotation_position="top right", line_color="#E87F2B")

    fig.update_layout(
            barmode='overlay',
            yaxis_title_text='Absolute Frequency',
            xaxis_title_text='Sample Mean',
            title=dict(
                text='<b><span style="font-size: 10pt">Sample Mean Distribution</span><br><span style="font-size: 8pt">Sample Size: '+str(sample_size)+' | Number of Samples: 100000</span> <span style="font-size:5">Data: Biased Die</span></b>',   # GEÄNDERT – Original: 'Sample Size: 20' / 'Data: UCIML Repo: Adult; variable: age'
            ),
        )
    fig.show()
```
